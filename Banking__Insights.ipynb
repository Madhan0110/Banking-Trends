```python
query61 = ("""

          SELECT BA.ACCOUNT_NUMBER,   BCM.CUSTOMER_MESSAGE,
          CASE
              WHEN BA.ACCOUNT_TYPE = "SAVINGS" AND BCM.EVENT = "Adhoc"
              THEN "Adhoc"
          END AS EVENTS
          FROM B_ACCOUNT BA CROSS JOIN B_CUSTOMER_MESSAGES BCM
          WHERE BA.ACCOUNT_TYPE = "SAVINGS" AND BCM.EVENT = "Adhoc"
          GROUP BY BA.ACCOUNT_NUMBER, BCM.EVENT, BCM.CUSTOMER_MESSAGE
          LIMIT 8;

""")
result61 = pd.read_sql_query (query61, conn)
print(tabulate(result61, headers= "keys", tablefmt="grid"))


```

```python
query_81 = ("""

            SELECT
                TABLE_8a.*, BAR.LINKING_ACCOUNT_NUMBER,
                BAR.ACCOUNT_TYPE AS LINKING_ACCOUNT_TYPE
              FROM (
                   SELECT BA.*, BT.TRANSACTION_AMOUNT
                   FROM
                      B_ACCOUNT BA LEFT JOIN B_ACCOUNT_TRANSACTION BT
                      ON BA.ACCOUNT_NUMBER = BT.ACCOUNT_NUMBER

                    ) AS TABLE_8a

              LEFT JOIN B_ACCOUNT_RELATIONSHIP_DETAILS BAR
               ON TABLE_8a.ACCOUNT_NUMBER = BAR.ACCOUNT_NUMBER

               limit 15 ;

""")

result_81 = pd.read_sql_query(query_81, conn)
print(tabulate(result_81, headers='keys', tablefmt='grid'))
```
